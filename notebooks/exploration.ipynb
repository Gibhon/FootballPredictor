```python
import pandas as pd
from pathlib import Path

data_path = Path.cwd().parent / "data"
raw_data_path = data_path / "raw" / "games.csv"
raw_data = pd.read_csv(raw_data_path)

print(raw_data.head(10))
```

```python
distinct_leagues = raw_data["competition_type"].unique()
distinct_leagues
```

```python
league_only_data = raw_data[raw_data["competition_type"] == "domestic_league"]
# print(league_only_data.info())
# print(league_only_data.isna().sum())
print(league_only_data.shape)
print(league_only_data.info())
print(league_only_data.describe())
```

```python
# Select your specific subset of columns and sum the missing values
null_counts = league_only_data[['season', 'round', 'date', 'home_club_name', 'competition_type','away_club_goals', 'home_club_goals']].isna().sum()

print(null_counts)
```
