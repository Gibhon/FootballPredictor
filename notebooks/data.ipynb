```python
import pandas as pd

from pathlib import Path

# 1. Define the root directory path
data_path = Path(__file__).resolve().parent / "data_multi_source"

# 2. Find ALL files ending in .csv in the folder and all subfolders (using rglob)
# Use "*.xlsx" for Excel files
all_files = list(data_path.rglob("*.csv"))

# 3. Read and combine them into a single list of DataFrames
df_list = []
for file in all_files:
    try:
        # Read the file
        df = pd.read_csv(file)

        df_list.append(df)
    except Exception as e:
        print(f"Skipping corrupted or empty file {file.name}: {e}")

# 4. Concatenate all DataFrames at once (much faster than combining inside the loop)
if df_list:
    combined_df = pd.concat(df_list, ignore_index=True)
    combined_df.to_csv("./big_data.csv", index=False)

    print(f"Combined Shape: {combined_df.shape}")
else:
    print("No valid data files found.")
```

```python
import pandas as pd

from pathlib import Path

# 1. Define the root directory path
domestic_path = Path.cwd().parent / "data_multi_source" / "domestic"

# 2. Find ALL files ending in .csv in the folder and all subfolders (using rglob)
# Use "*.xlsx" for Excel files
all_files = list(domestic_path.rglob("*.csv"))

# 3. Read and combine them into a single list of DataFrames
df_list = []
for file in all_files:
    try:
        # Read the file
        df = pd.read_csv(file)

        df_list.append(df)
    except Exception as e:
        print(f"Skipping corrupted or empty file {file.name}: {e}")
```

```python
print([df.columns for df in df_list])
```

```python
import math

n = 5
# Calculate base chunk sizes
avg = len(df_list) / n
split_lists = []
last = 0.0

while last < len(df_list):
    # This rounding logic ensures items are distributed as evenly as possible
    split_lists.append(df_list[int(last) : int(last + avg)])
    last += avg

# print the 5 lists
for i, sublist in enumerate(split_lists):
    common_columns = list(
        reduce(lambda x, y: x.intersection(y), [df.columns for df in sublist])
    )
    print(f"List{i}: {len(common_columns)}")

common_columns = list(
    reduce(lambda x, y: x.intersection(y), [df.columns for df in df_list])
)


all_columns = list(reduce(lambda x, y: x.union(y), [set(df.columns) for df in df_list]))
print(len(common_columns))

print("All Unique Columns:", len(all_columns))

# 4. Concatenate all DataFrames at once (much faster than combining inside the loop)
if df_list:
    combined_df = pd.concat(df_list, ignore_index=True)
    combined_df.to_csv("./big_data.csv", index=False)

    print(f"Combined Shape: {combined_df.shape}")
else:
    print("No valid data files found.")
combined_df.shape
```
